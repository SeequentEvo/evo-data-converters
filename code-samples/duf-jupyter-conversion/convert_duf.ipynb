# DUF Converter

This notebook shows how you can sign in and publish geoscience objects from a DUF file to your chosen Evo workspace.

**Important:** This notebook requires Python 3.10, 3.11, or 3.12 (not 3.14+) due to evo.notebooks dependencies.

In the first code cell we create a `ServiceManagerWidget` which will open a browser window and ask you to sign in.

Once signed in, a widget will be displayed to allow you to select your organisation and an Evo workspace.

__Required:__ In the first code cell, replace `"your-client-id"` with your Evo app client ID before running the cell.

**Note:** This notebook assumes your app's redirect URL is `http://localhost:3000/signin-callback` (the default). If your app uses a different callback URL, add the `redirect_url` parameter:

```python
manager = await ServiceManagerWidget.with_auth_code(
    client_id="your-client-id",
    redirect_url="http://your-custom-url/callback"
).login()
```

````python
import warnings

from evo.notebooks import ServiceManagerWidget

# Suppress all warnings
warnings.simplefilter("ignore")

manager = await ServiceManagerWidget.with_auth_code(client_id="your-client-id").login()
````

## Convert and Publish DUF File

Run the cell below to display the conversion widget. It will allow you to:
1. Select a `.duf` file from your local filesystem
2. Enter an EPSG code for the coordinate reference system
3. Optionally specify an object path in the Evo workspace
4. Convert and publish the DUF file to your selected workspace

````python
from helpers.widgets import create_duf_widget

ui = await create_duf_widget(manager)
display(ui)
````
